```python
# Load pre-trained YOLOv5 model (use YOLOv5s, YOLOv5m, YOLOv5l, or YOLOv5x as needed)
model = torch.hub.load('ultralytics/yolov5', 'yolov5s')  # You can switch 'yolov5s' with 'yolov5m', 'yolov5l', or 'yolov5x'

# Path to the directory containing the images from the Chemed Telegram Channel
image_folder = '../Data-Warehouse-and-Object-Detection/media'

save_dir='../Data-Warehouse-and-Object-Detection/YOLO/detected_results'
# Create the save directory using save_dir instead of base_save_dir
os.makedirs(save_dir, exist_ok=True)

import glob

# Define a list to store the paths of the detected images for later display
detected_dir = []

# Loop through the images in the folder and run object detection
for img_name in os.listdir(image_folder):
    img_path = os.path.join(image_folder, img_name)

    # Check if the file is an image
    if not (img_name.endswith(".png") or img_name.endswith(".jpg") or img_name.endswith(".jpeg")):
        print(f"Skipping {img_name} - not an image file")
        continue

    # Read the image using OpenCV
    img = cv2.imread(img_path)

    # Run object detection
    results = model(img)

    # Show the results
    # results.show()  # Show detected objects on the image
    results.save(save_dir=save_dir)  # Save detection results to a directory
clear_output()
```

```python
DIR = glob.glob(f"{save_dir}*/")
for sub_dir in DIR:
    detected_img_path = os.path.join(sub_dir, 'image0.jpg')
    detected_dir.append(detected_img_path)

# Debugging: print out the first few image paths
print("Sample detected images to be displayed:")
print(detected_dir[:5])
```

```python
import matplotlib
%matplotlib inline

import matplotlib.image as mpimg

# Check if detected_dir is not empty and contains valid images
if len(detected_dir) == 0:
    print("No images found in the detected directories.")
else:
    # print(f"Images found: {detected_dir}")

    # Display sample images using subplots with larger figure size
    fig = plt.figure(figsize=(30, 40))  # Increase the figure size for larger images
    rows = 7
    cols = 3

    # Ensure there are at least 21 images to display
    images_to_display = detected_dir[:21]

    for i, ax in enumerate(fig.subplots(rows, cols).flatten()):
        try:
            img_path = images_to_display[i]

            # Check if the image file exists
            if not os.path.exists(img_path):
                print(f"Image not found: {img_path}")
                continue

            img = mpimg.imread(img_path)
            ax.imshow(img)
            ax.axis('off')  # Turn off axis for clearer view

        except Exception as e:
            print(f"Error loading image {img_path}: {e}")

    fig.tight_layout(pad=2.0)  # Adjust the padding for more space between subplots
    plt.show()
```

```python
# Ensure the output directory exists
output_folder = '../Data-Warehouse-and-Object-Detection/YOLO'
os.makedirs(output_folder, exist_ok=True)  # Create if it doesn't exist

# Process results to extract bounding box, class labels, and confidence scores
for sub_dir in DIR:
    # Find all image files within the subdirectory
    image_files = glob.glob(os.path.join(sub_dir, '*.jpg'))  # Search for all .jpg files

    # Skip subdirectory if no images are found
    if not image_files:
        print(f"No images found in {sub_dir}. Skipping...")
        continue

    # Iterate through each image file found
    for img_path in image_files:
        # Run object detection
        results = model(img_path)

        # Extract data from results
        detections = results.pandas().xyxy[0]  # Pandas dataframe of detection results

        # Get image name from the path
        img_name = os.path.basename(img_path)

        print(f"Detections for {img_name}:")
        print(detections[['name', 'confidence', 'xmin', 'ymin', 'xmax', 'ymax']])

        # Save detections to CSV file
        output_path = os.path.join(output_folder, f'{os.path.splitext(img_name)[0]}_detections.csv')
        detections[['name', 'confidence', 'xmin', 'ymin', 'xmax', 'ymax']].to_csv(output_path, index=False)
        print(f"Saved detection results to {output_path}")
```

For each a series of image, the detection results include objects with their **corresponding labels**, **confidence scores**, and **bounding box coordinates** (`xmin`, `ymin`, `xmax`, `ymax`).

**Key points:**

* Multiple detections were made for each image, with objects such as people, bottles, chairs, and other items identified.
* Confidence levels for detections varied, indicating varying degrees of certainty in the predictions.
* Some entries resulted in empty detection outputs, implying no objects were found in those instances.

The results were saved as CSV files in a specified directory for further analysis.
